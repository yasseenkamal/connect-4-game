```python
import os, sys
cwd = os.getcwd()
print(" Current working directory:", cwd)

candidates = [
    os.path.join(cwd,    'src'),
    os.path.join(cwd, '..','src'),
]
found = []
for p in candidates:
    if os.path.isdir(p):
        found.append(os.path.abspath(p))

if not found:
    raise FileNotFoundError(
        " Could not find a 'src/' directory. "
        f"Tried these: {candidates}"
    )

src_path = found[0]
print("Found src/ at:", src_path)

if src_path not in sys.path:
    sys.path.insert(0, src_path)
print(" First entries of sys.path:", sys.path[:3])
print(" Contents of src/:", os.listdir(src_path))
```
