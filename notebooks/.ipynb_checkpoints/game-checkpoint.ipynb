```python
# Cell 1: Auto‑detect and add src/ to sys.path

import os, sys

# 1. Where am I right now?
cwd = os.getcwd()
print("👣 Current working directory:", cwd)

# 2. Look for src/ either here or one level up
candidates = [
    os.path.join(cwd,    'src'),
    os.path.join(cwd, '..','src'),
]
found = []
for p in candidates:
    if os.path.isdir(p):
        found.append(os.path.abspath(p))

if not found:
    raise FileNotFoundError(
        "❌ Could not find a 'src/' directory. "
        f"Tried these: {candidates}"
    )

# 3. Use the first match
src_path = found[0]
print("✅ Found src/ at:", src_path)

# 4. Insert into Python path (if not already there)
if src_path not in sys.path:
    sys.path.insert(0, src_path)
print("🔍 First entries of sys.path:", sys.path[:3])

# 5. (Optional) Show contents of src/ so you know game.py really is there
print("📂 Contents of src/:", os.listdir(src_path))
```
